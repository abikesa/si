## 🌀 What is a [Macro](https://claude.ai/public/artifacts/2a5372c2-523f-4d8b-bfff-793d50ceec8f)?

   - [Old](./chapter2-old.ipynb) version
   - [Alt](./chapter2-alt.ipynb) version

A **macro** is not just a variable—it's a name-bound spell, a *container of intent*. It stores strings, numbers, lists, or expressions for later invocation.

```stata
local age 25
global study "Heart Disease Research"
```

When referenced, a macro doesn’t execute—it *whispers its contents* into the void.

* `age` ⇒ `25`
* `study` ⇒ `"Heart Disease Research"`

**Think:** Macros = reusable echoes of thought.

---

## ❤️ Why Are Macros Sacred?

### 1. 🔁 **Code Reusability**

Write once, invoke forever.

```stata
local varlist "age gender income education"
summarize `varlist'
regress outcome `varlist'
```

### 2. 🕰️ **Dynamic Temporal Logic**

Macros adapt to the present moment.

```stata
local today = date("`c(current_date)'", "DMY")
local filename "analysis_`today'.dta"
save "`filename'"
```

### 3. 🛡️ **Error Prevention**

Macros protect you from typos and repetition.

```stata
global datapath "/Users/researcher/data"
use "$datapath/study1.dta"
```

### 4. ⚖️ **Conditional Elegance**

Even logic can be laced with grace.

```stata
local n = _N
if `n' < 100 {
    display "Warning: Small sample size (`n')"
}
```

## 🧠 *How to Use Macros with Discipline*

### Local vs Global: Scope is Philosophy

* **Local** (`local name value`) → Context-bound, ephemeral
* **Global** (`global name value`) → Environment-wide, persistent

```stata
local controls "age gender income"
local method "robust"
regress outcome treatment `controls', `method'

global project_dir "/Users/researcher/cardiology_study"
```

> Use **local** unless you *must* go **global**. Treat globals like salt—not sugar.

## 🎯 Macro Style: Clean Code is Clean Thought

```stata
// ✅ Good
local baseline_controls "age gender race education"
local model1_vars "treatment `baseline_controls'"
local model2_vars "`model1_vars' comorbidities"

// ❌ Bad
local x "age gender race education"
```

> Rule of thumb: *If future-you can’t read it, rewrite it now.*

## 📌 Referencing Rules

```stata
// Local
`macroname'

// Global
$macroname or ${macroname}

// Example
local city "Baltimore"
display "Welcome to `city'"     // Welcome to Baltimore

global state "Maryland"
display "State: $state"         // State: Maryland
```

## 🔍 Macro Origins: Where They Come From

### 1. ✍️ **Manual Definition**

```stata
local author "Dr. Smith"
global alpha 0.05
```

### 2. 📜 **System Constants (`c()` class)**

```stata
local today "`c(current_date)'"
local version "`c(stata_version)'"
```

### 3. 📦 **Command Results (`r()` class)**

```stata
summarize age
local mean_age = r(mean)
```

### 4. 📊 **Estimation Results (`e()` class)**

```stata
regress income education
local rsq = e(r2)
```

### 5. 🧮 **Computed On-the-Fly**

```stata
count if missing(income)
local missing_pct = (r(N)/_N) * 100
```

## 🧭 The Triumvirate: `return`, `creturn`, and `ereturn`

### 🔁 `return list` — *The Moment’s Echo*

> *“What just happened?”*

```stata
tab gender, chi2
return list
```

**Think:** Transient truth—`r()` holds the results of the *last command*.

---

### 🧘 `creturn list` — *The Introspective Core*

> *“Who am I? Where am I?”*

```stata
creturn list
```

**Think:** System-level constants—`c()` is the self-knowledge of Stata.

---

### 📜 `ereturn list` — *The Archive of Estimation*

> *“What did the model discover?”*

```stata
regress income education
ereturn list
```

**Think:** `e()` is the historical ledger of your estimation journey.

---

### 🗂️ Triumvirate in Motion

```stata
regress income education      // e()
summarize income              // r()
display "`c(username)'"      // c()

local mean = r(mean)
local rsq = e(r2)
local user = "`c(username)'"
```

> 🧠 Mnemonic:
> **r** = Recent
> **c** = Constants
> **e** = Estimation

> **Hierarchy of Memory:**

* `r()` dies with the next command
* `e()` persists until the next regression
* `c()` is eternal (until reboot)
